#Linear Regression


```python
class LinearRegression:
  def __init__(self,x,y):
    self.x=x;
    self.y=y;
    self.x_avg=sum(x)/len(x)
    self.y_avg=sum(y)/len(y)

  def coef(self):
    num=0
    den=0
    for i in range(len(self.x)):
      num+=(self.x[i]-self.x_avg)*(self.y[i]-self.y_avg)
      den+=(self.x[i]-self.x_avg)*(self.x[i]-self.x_avg)
    return num/den

  def intercept(self):
    b=self.y_avg-self.coef()*self.x_avg
    return b

  def fit(self):
    print(self.coef())
    print(self.intercept())

  def predict(self,x):
    return self.coef()*x+self.intercept()
```
