```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import make_classification
X,y=make_classification(
    n_samples=40,
    n_features=3,
    n_informative=3,
    n_redundant=0,
    n_classes=2,
    random_state=23
)

df=pd.DataFrame(X,columns=['Feature 1','Feature 2','Feature 3'])
df['target']=y
```

```python
df.head()
```

```python
import plotly.express as px
fig=px.scatter_3d(
    df,
    x='Feature 1',
    y='Feature 2',
    z='Feature 3',
    color='target'
)
fig.show()
```
